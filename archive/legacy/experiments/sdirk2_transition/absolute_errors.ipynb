# 步长与时间变化下的绝对误差

参照 `sdirk2_fixed_convergence_table.tex`，使用涡量**绝对** $L^2$ 误差：
$$E_\omega(\tau,t)=\sqrt{h_xh_y\sum_{i,j}(\omega_{ij}(\tau,t)-\omega_{{\rm ref},ij}(t))^2}.$$

当前算例：$\nu=0.2$、$N=256$、外力 $\cos x$、原多模态初值，比较 SDIRK2 和 mr-SAV 的 $\gamma=5,100,500,1000$。表格按实际步长与物理时间组织，不显示相对误差或百分数。

主表采用 $t=1.5,3,4.5,6$ 的共同时间节点；密集过渡步长另列终点表。时间曲线全部取方法和参考解的真实公共节点，没有插值，也没有为输出而缩短时间步。

## 读图结论与验证范围

小步长下三组误差接近，误差可以随物理解演化而减小；较大步长下误差在后期转为快速增长。gamma=5 的明显优势主要发生在该后期增长阶段，不能从终点表推出整个时间区间都有同等幅度的优势。

补算的参考解和所有方法的末状态均与此前保存结果逐点一致。两张 LaTeX 表编译通过。`stopped` 表示在该时刻前已经触发预设停止阈值，不把停止后缺失结果伪造为数值 NaN。完整说明见 `ABSOLUTE_ERRORS.md`。
